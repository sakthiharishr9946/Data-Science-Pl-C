```python
import pandas as pd

students=pd.read_excel("C:/Users/harish/Downloads/Merge_Join_Concat_Dataset_50_Records.xlsx",sheet_name="students")

marks=pd.read_excel("C:/Users/harish/Downloads/Merge_Join_Concat_Dataset_50_Records.xlsx",sheet_name="marks")

contact=pd.read_excel("C:/Users/harish/Downloads/Merge_Join_Concat_Dataset_50_Records.xlsx",sheet_name="contact")
```

```python
# 1. INNER MERGE

#Returns only Matching Student_IDs from both DataFrames.

inner_df=pd.merge(
    students,
    marks,
    on="Student_ID",
    how="inner"
)

print(inner_df.head())
print(inner_df.shape)
```

```python
# 2.LEFT MERGE
left_df=pd.merge(
    students,
    marks,
    on="Student_ID",
    how="left"
)

print(left_df.head())
print(left_df.shape)

#Even if marks are missing
```

```python
# 3.RIGHT MERGE
right_df=pd.merge(
    students,
    marks,
    on="Student_ID",
    how="right"
)

print(right_df.head())
print(right_df.shape)
```

```python
# 4. OUTER MERGE
# Keeps everything from both DataFrames.
outer_df=pd.merge(
    students,
    marks,
    on="Student_ID",
    how="outer"
)

print(outer_df.head())
print(outer_df.shape)
```

```python
# 5.

#Combine Students + Marks + Contact.

student_complete=pd.merge(
    students,
    marks,
    on="Student_ID",
    how="left"
)

student_complete=pd.merge(
    student_complete,
    contact,
    on="Student_ID",
    how="left"
)

print(student_complete.head())
```

```python
# 6. CONCAT ROW-WISE

#Split Students into two parts.

part1=students.iloc[:25]
part2=students.iloc[25:]

combined_rows=pd.concat([part1,part2])

print(combined_rows.head())
print(combined_rows.shape)
```

```python
# 7.CONCAT COLUMN-WISE

student_basic=students[['Student_ID','Name']]
student_marks=marks[['Python']]
combined_cols=pd.concat(
    [student_basic,student_marks],
    axis=1
)

print(combined_cols.head())
print(combined_cols.shape)
```

```python
# 8. JOIN()

#join() works using indexes.

students_index=students.set_index("Student_ID")
contact_index=contact.set_index("Student_ID")

joined_df=students_index.join(contact_index)

print(joined_df.tail(10))
```

## Pivot Table

Pivot Table is used to summarize and analyze data by grouping rows and applying aggregate functions such as sum, mean, count, max, min, etc.

```python
# Step 1: Load Dataset

import pandas as pd

df5=pd.read_csv("C:/Users/harish/Downloads/people-10000.csv")

print(df5.head())
```

```python
# Step 2: Create Age Column
df5['Date of birth']=pd.to_datetime(df5['Date of birth'])

today=pd.Timestamp.today()

df5['Age']=(
    (today-df5['Date of birth']).dt.days//365
)

print(df5[['First Name','Sex','Age']].head())
```

--if cannot upload files directly in colab
* from google.colab import files

```python
# 1. Basic Pivot Table

# Average age by gender.

pivot1=pd.pivot_table(
    df5,
    values='Age',
    index='Sex'
)

print(pivot1)
```

```python
# 2.Pivot Table with aggfunc='mean'

pivot2=pd.pivot_table(
    df5,
    values='Age',
    index='Sex',
    aggfunc='mean'
)

print(pivot2)
```

```python
# 3.Pivot Table with aggfunc='sum'

pivot3=pd.pivot_table(
    df5,
    values='Age',
    index='Sex',
    aggfunc='sum'
)

print(pivot3)
```

```python
# 4.Pivot Table with aggfunc='count'

pivot4=pd.pivot_table(
    df5,
    values='Age',
    index='Sex',
    aggfunc='count'
)

print(pivot4)
```

```python
#5. Multiple Aggregations.

pivot5=pd.pivot_table(
    df5,
    values='Age',
    index='Sex',
    aggfunc=['mean','max','min']
)

print(pivot5)
```

```python
#6. Pivot Table with margins=True

# Adds overall average

pivot6=pd.pivot_table(
    df5,
    values='Age',
    index='Sex',
    aggfunc='mean',
    margins=True
)

print(pivot6)
```

```python
#7 .Group by Job Title and Gender

# Average age by Job Title and Gender.

pivot7=pd.pivot_table(
    df5,
    values='Age',
    index='Job Title',
    columns='Sex',
    aggfunc='mean'
)
print(pivot7)
```

```python
# 8. 

# Count People by Job and Gender

pivot8=pd.pivot_table(
    df5,
    values='User Id',
    index='Job Title',
    columns='Sex',
    aggfunc='count',
    fill_value=0
)

print(pivot8)

#nan fill value with 0
```
