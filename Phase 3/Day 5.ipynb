```python
#Figure and Axes
#These are the most important concepts.
#Figure - The entire chart window.
#Axes-The area where the graph is drawn.

import matplotlib.pyplot as plt


fig,ax=plt.subplots()

x=[1,2,3]
y=[10,20,30]

ax.plot(x,y)

#Adding a Title
plt.title("Monthly Sales")

#adding X and Y Lables
plt.xlabel("Month")
plt.ylabel("Sales")

plt.plot(x,y,color='green')
plt.plot(x,y,linestyle='-.')
plt.show()

#Styles:

#'-' Solid
#'--' Dashed
#':' Dotted
#'-.' Dash Dot
```

## Box Plot

A box plot is used to visualize the distribution of numerical data and identify outliers.


##### Components of a Box Plot:
* **Minimum**: Smallest Value (excluding outliers)
* **Q1(25%)**: First Quartile
* **Median(50%)**: Middle Value
* **Q3(75%)**: Third Quartile
* **Maximum**: Largest Value (excluding outliers)
* **Outliers**: Values far from the rest of the data,shown as individual points.

```python
import matplotlib.pyplot as plt

salary=[25000,28000,30000,32000,35000,37000,40000,45000,100000]

plt.boxplot(salary)
plt.title("Salary Distribution")
plt.ylabel("Salary")
```

## Heatmap

A Heatmap uses colors to represent valuesin a table or matrix

In data Analysis,heatmaps are commonly used to visualize correlations between numerical columns.

Correlation Values:

value meaning:

* +1 - Strong Positive relationship
* 0 - No Relationship
* -1 - Strong Negative Relationship



```python
import seaborn as sns
import matplotlib.pyplot as plt

tips=sns.load_dataset("tips")

#Correlation matrix
corr=tips.corr(numeric_only=True)
sns.heatmap(corr,annot=True)

plt.title("Histogram")

               
```

```python
# Histogram Using Seaborn

sns.histplot(data=tips,x="total_bill",bins=30)

plt.title("Distribution of Total Bill")
plt.show()

#sns.histplot()
#Creates histogram.

```

## KDE Plot

A KDE Plot is a smooth version of histogram,

instead of bars, it shows a smooth curve representing data distribution,

when to use kde?

understanding data distribution identifying peaks comparing distributions statistical analysis.

```python
sns.kdeplot(data=tips,x='total_bill')
plt.title("KDE Plot")
plt.show()
```

```python
#Histogram + KDE Plot
#Most Commonly used in Data Science

sns.histplot(data=tips,x="total_bill",kde=True)
plt.title("Histogram with KDE")
plt.show()

#kde=True
#Adds a smooth density curve on top of the histogram
```

## Subplots

what is subplots?

subplots allow multiple charts in a single figure.

syntax: plt.subplot(rows,columns,position)

```python
#2 charts

import matplotlib.pyplot as plt
plt.figure(figsize=(8,4))

# First Plot
plt.subplot(1,2,2)
plt.plot([1,2,3], [10,20,30])
plt.title("Line Chart")

# Second Plot
plt.subplot(1,2,1)
plt.bar(['A','B','C'], [5,8,6])
plt.title("Bar Chart")
plt.show()

#Example: Four Charts
plt.figure(figsize=(10,8))
plt.subplot(2,2,1)
plt.plot([1,2,3], [10,20,30])
plt.title("Line")
plt.subplot(2,2,2)
plt.bar(['A','B','C'], [5,8,6])
plt.title("Bar")
plt.subplot(2,2,3)
plt.scatter([1,2,3], [4,8,6])
plt.title("Scatter")
plt.subplot(2,2,4)
plt.pie([30,40,30], labels=['A','B','C'])
plt.title("Pie")
plt.show()
```
