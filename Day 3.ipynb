## Series

```python
import pandas as pd
a=pd.Series(['A','B','C','D'])
print(a)
print(a.values)
print(a.index)
print(a.dtype)
```

```python
#Accessing Elements
print(a[1])
print(a.iloc[1])
```

```python
#Slicing
print(a[1:3])
print(a.loc[1:3])
print(a.iloc[2:5])
```

```python
arr=pd.Series([1,2,3,4,56],index=['A','B','C','D','E'])
print(arr)
print(arr['E'])
```

```python
#Vectorized operation - no loop required
print(arr+5)
```

```python
print(arr.loc['A'])
```

### Differnce Between loc() and iloc()
* loc() --it accesses element based on index value
* iloc() --it accesses element based on position

```python
print(arr.sum())
print(arr.std())
print(arr.max())
print(arr.min())
print(arr.mean())
print(arr.median())
```

```python
print(arr[arr>30])
```

```python
#count values
print(len(arr))
```

```python
#it return individual value count
print(arr.value_counts())
```

## DataFrame

```python
import pandas as pd
#method1 : from dictionary

data={
    'name':['mercury','venus','earth'],
    'rollno':[12,24,36],
    'age':[20,19,22]
}

df=pd.DataFrame(data)
```

```python
print(df)
```

```python
#from Series to DataFrame
s1=pd.Series([1,2,3,4,5])
s2=pd.Series([5.6,9.54,9.4,54.6,6])
s3=pd.Series([True,False,True,False])
df=pd.DataFrame({
    'sno':s1,
    'cgpa':s2,
    'attendance':s3
})
print(df)
```

```python
#List of List
data=[['Jupiter',22,98],
      ['Saturn',20,89],
      ['Uranus',20,97]]
df=pd.DataFrame(data,columns=["Name","Age","Mark1"])
print(df)
```

```python
#List of Dictionary
data=[
    {"Name":"Jupiter","Course":"DS","Grade":"A"},
    {"Name":"Saturn","Course":"CSDA","Grade":"B"},
    {"Name":"Uranus","Course":"AIDS","Grade":"A+"}
]

df=pd.DataFrame(data,index=["std1","std2","std3"])
print(df)

#Adding one extra columns
df['Marks']=[98,87,67]
print("\n",df)
```

```python
#Adding One Extra Row
std4=pd.DataFrame([{"Name":"Neptune","Course":"DS","Grade":"O","Marks":78}])
df=pd.concat([df,std4])
print(df)
```

```python
std4=pd.DataFrame({"Name":["Neptune"],"Course":["DS"],"Grade":["O"],"Marks":[78]})
df=pd.concat([df,std4],ignore_index=True)
print(df)
```

```python
#Adding Multiple Extra Rows
students=pd.DataFrame([{"Name":"Mercury","Course":"AIML","Grade":"O+","Marks":92},
                       {"Name":"Venus","Course":"CSDA","Grade":"A++","Marks":90}],)
df=pd.concat([df,students])
print(df)
```

```python
print(df.Course)
print("\n",df[['Grade','Marks']])
```

```python
# Shape Attribute
print(df.shape[0])
# Find No of Columns
print(df.shape[1])
#DType attribute
#Used to check data type of all columns
print(df.dtypes)
#info
print(df.info())
print(df.describe())
```

```python
#Select Specific Row
print(df.loc[['std1','std3']])
print(df.loc[1])
```

```python
# Update Value
df.loc['std3','Grade']='B++'
print(df)
```

```python
df.drop("Grade",axis=1,inplace=True)
print(df)
```

```python
df=df.drop('std3',axis=0)
print(df)
```

```python
df.sort_values('Marks',inplace=True)
print(df)
```

```python
df.sort_values('Marks',ascending=False,inplace=True)
print(df)
```

```python
# Statistical Function for Particular Column
print(df['Marks'].max())
```
