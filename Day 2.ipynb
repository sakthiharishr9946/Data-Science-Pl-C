## Statistical Function

--Uses
* Data Analysis
* Find Average
* Spread of Data
* Find Relationship between Variables

1. **Mean()** -- Find the Average Values
2. **Median()** -- Finds the Middle Value after sorting the array
3. **std()** (standard deviation) --How far Data are spread from mean
   * Small std -- data close to mean
   * Large std -- data spread wide
  
4. **Correlation** -- 
   * Positive
   * Negative
   * No
  
5. **Percentile** --

```python
# Statistical Function
import numpy as np
p=np.array([10,2,30,40,5])
print(np.mean(p))
print(np.median(p))
print(np.std(p))
print(np.max(p))
print(np.min(p))
print(np.percentile(p,100))
print(np.var(p))
print(np.argmin(p))
```
